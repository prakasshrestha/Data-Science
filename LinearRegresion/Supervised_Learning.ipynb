### Algorithms

* Classification goal is to assign input data to predefined categories or classes. Algorithms like K-Nearest Neighbors(KNN), Logistic Regression, Decision Trees, Support Vector Machines, and Neural Netwroks are commanly use for classification task.

* Regression goal is to predict a numerical value based on input features. algorithms like linear Regression, polynomial Regression, and various type fo regression trees are used for regression task.

## Regression 
* Regression is a statistical method that helps us understand and predict the relationship between variable 

* Describes how one variable (dependent variable) changes as another variable(independent variable) changes.

* Independent variable: that are used to predict or explain the changes in the dependent variable(X).

* Dependent variable: We are trying to predict or explain(Y).

For example : predicting salary based on years of experience, predicting examp score based on study hours,predicting resale price based on vehicle age.

## Linear Regression

Equation of linear regression : $ Y = mx+ b $
* Y represents the dependent variable 
* X represents the independetn variable. 
* m is the slope of the line(how much y changes for a unit change in X).
* b is the intercept ( the value of y when x is 0).
